```python
import gym
```

# Create environment

```python
env = gym.make('CartPole-v1')
```

# Run episodes

```python
for episode in range(5):
    state, info = env.reset() # Update for gymnasium
    done = False
    truncated = False

    while not done and not truncated:
        # env.render() # Removed as render_mode is set in gym.make
        action = env.action_space.sample()
        state, reward, done, truncated, info = env.step(action)
        print("Reward:", reward)

env.close()
```
